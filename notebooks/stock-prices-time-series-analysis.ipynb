#### custom calender

```python
from pandas.tseries.holiday import AbstractHolidayCalendar, Holiday, nearest_workday

class MyBirthdayCalendar(AbstractHolidayCalendar):

    rules = [
        Holiday("Gichuki's Birthday", month=1, day=31, observance=nearest_workday),
        
    ]

my_birthday_calendar = CustomBusinessDay(calendar=MyBirthdayCalendar())
my_birthday_calendar
```

```python
# Example usage of the custom calendar
pd.date_range(start='2025-01-20', end='2025-02-10', freq=my_birthday_calendar)
```

### 📊 Data Dictionary: Nvidia Financial Report

This dataset contains key financial metrics for Nvidia Corporation across multiple fiscal years. The data is structured with **rows representing financial line items** (e.g., Revenue, Gross Profit) and **columns representing time periods**. Below is an explanation of each column:

- **Fiscal Year**: The name of the financial metric or KPI (e.g., Revenue, Net Income, etc.).
- **TTM (Trailing Twelve Months)**: Represents cumulative financial performance for the most recent 12-month period leading up to the report (not necessarily aligned with a fiscal year). Useful for evaluating up-to-date performance.
- **FY 2025**: Projected or actual financial data for Nvidia’s Fiscal Year 2025. This fiscal year **ends in Jan 2025**, meaning it covers **February 2024 to January 2025**.
- **FY 2024**: Actual financials for Fiscal Year 2024 (Feb 2023 – Jan 2024).
- **FY 2023**: Actual financials for Fiscal Year 2023 (Feb 2022 – Jan 2023).
- **FY 2022**: Actual financials for Fiscal Year 2022 (Feb 2021 – Jan 2022).
- **FY 2021**: Actual financials for Fiscal Year 2021 (Feb 2020 – Jan 2021).

> 🧠 **Note**: Nvidia's fiscal years do **not** align with calendar years. Each FY is labeled by the **year in which it ends** (e.g., FY 2024 ends in January 2024, but starts in February 2023).

The actual period end dates (e.g., `Jan 28, 2024`) are shown in the first row of the dataset under each FY column for reference.
